```python
# -*- coding: utf-8 -*-
"""preprocess_letter box.ipynb

Automatically generated by Colab.

Original file is located at
    https://colab.research.google.com/drive/16mMW2klJOd4uSAOc9ZHknFyj990iIuy6
"""

import os
import cv2
import numpy as np
from tqdm import tqdm
import random
import shutil

# Paths
INPUT_DIR = "/content/drive/MyDrive/Major_Project/Images900/ORIGINAL_3_CLASS/speed_breaker"
OUTPUT_DIR = "/content/drive/MyDrive/Major_Project/Images900/SpeedBreaker_Preprocessed"

os.makedirs(OUTPUT_DIR, exist_ok=True)

# --- SAFE AUGMENTATION (NO HORIZONTAL FLIP) ---
def augment(img):
    # Rotation ±10 degrees
    angle = random.uniform(-10, 10)
    h, w = img.shape[:2]
    M = cv2.getRotationMatrix2D((w/2, h/2), angle, 1)
    img = cv2.warpAffine(img, M, (w, h), borderValue=(255,255,255))

    # Brightness 0.9 to 1.1
    brightness = random.uniform(0.9, 1.1)
    img = np.clip(img * brightness, 0, 255).astype(np.uint8)

    return img


# ================= PROCESS SCHOOL ZONE IMAGES =================
NUM_IMAGES_TARGET = 900

print(f"Processing SpeedBreaker images from: {INPUT_DIR}")

# Get all images from the input directory
images = [os.path.join(INPUT_DIR, f) for f in os.listdir(INPUT_DIR)
          if f.lower().endswith(('.png', '.jpg', '.jpeg'))]

print(f"Found {len(images)} original images")

# Load all original images
processed = []

for img_path in tqdm(images, desc="Loading original images"):
    img = cv2.imread(img_path)
    if img is None:
        continue

    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    processed.append(img)

print(f"Successfully loaded {len(processed)} images")

# Augment to reach 900 images
if len(processed) < NUM_IMAGES_TARGET:
    print(f"Augmenting images from {len(processed)} to {NUM_IMAGES_TARGET}...")

    while len(processed) < NUM_IMAGES_TARGET:
        base_img = random.choice(processed[:len(images)])  # Augment from original images
        aug_img = augment(base_img)
        processed.append(aug_img)
else:
    print(f"Already have {len(processed)} images, selecting {NUM_IMAGES_TARGET}...")
    processed = processed[:NUM_IMAGES_TARGET]

# Save all processed images
print(f"Saving {len(processed)} images to {OUTPUT_DIR}...")

for i, img in enumerate(tqdm(processed, desc="Saving images")):
    img = cv2.cvtColor(img, cv2.COLOR_RGB2BGR)
    path = os.path.join(OUTPUT_DIR, f"Speed_Breaker_{i:04d}.jpg")
    cv2.imwrite(path, img)

print("\n✅ Preprocessing Completed!")
print(f"Total images saved: {len(processed)}")
print(f"Output directory: {OUTPUT_DIR}")
```
